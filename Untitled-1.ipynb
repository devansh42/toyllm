```python

def load_dataset(path) -> str:
    with open(path) as f:
        return "".join(f.readlines())
```

```python
text = load_dataset("/Users/devansh/go/src/github.com/devansh42/llm/script.txt")
text
```

```python
EOS = '>' # End of Sentence
PAD = 0 # Padding
EOW = '<' # End of Word
EOC = ']' # End of Chapter
PAR = '[' # Paragraph
person_titles = ['mr','miss','ms','mrs']
```

```python
"""
Normalises text by adding appropirate special characters
"""
def to_normalised_text(text:str)->list[str]:
    lines = text.splitlines()
    text =" ".join(lines)
    words =text.split(" ")
    out = []
    for word in words:
        w=""
        if word.endswith("."):
            _w = word[:len(word)-1]
            if _w.lower() in person_titles:
                w += word + EOW
            else:
                w+=_w + "." + EOW + EOS
        else:
            w+=word    
            w+=EOW
        w = w.lower()
        out.append(w)
    final_out = []
    for w in out:
        w= PAR if w == EOW else w
        final_out.append(w)
    return final_out 
            

    # return [SOW + word.lower() for word in words if word.isalnum()]
input_words = to_normalised_text(text)
input_words
```

```python

class BPETokenizer:
    def __init__(self, vocab_len:int):
        self.vocab_len = vocab_len
        self.vocab = dict()

    def initial_vocab(self, input:list[str]):
        for word in input:
            alphs = set(word)
            for c in alphs:
                self.vocab[c] = 1 + (0 if c not in self.vocab else self.vocab[c] )
    def train_vocab(self,input:list[str]):
        self.initial_vocab(input)
        i = len(self.vocab)
        
        while (i < self.vocab_len):
            freq = dict()
            for word in input: # This makes token pair at word boundary
                freq = self.pair_tokens(word,freq)
            if len(freq) == 0:
                print("no updates in pairing")
                break # no updates in pairing
            (pair, f) = self.max_pair(freq)
            input = [self.merge(pair,word) for word in input] # Merges happens at word boundary
            pre_l = len(self.vocab)
            self.vocab[pair] = f
            post_l = len(self.vocab)
            if pre_l == post_l:
                break # no more updates
            i+=1


    def merge(self, new_token:str, word : list[str]) -> list[str]:
        new_token_l = list(new_token)
        out = []
        i = 0
        while( i < len(word)):
            if i + 2 < len(word) and list("".join(word[i:i+2])) == new_token_l:
                out.append(new_token)
                i+=2
            else: 
                out.append(word[i])
                i+=1
        return out
    def pair_tokens(self,word: list[str],freq: dict[str,int]) -> dict[str,int]:
        tl = len(word) - 1
        for i in range(tl-1):
            pair = word[i] + word[i+1]
            
            freq[pair] = 1 if pair not in freq else freq[pair] + 1 # Check this statment for terminal character
            i+=1
        return freq
    def max_pair(self, pfreq: dict[str,int]) -> tuple[str,int]:
        ar = []
        ar = [(k,pfreq[k]) for k in pfreq]
        ar.sort( key = lambda pf : pf[1],reverse=True)
        return ar[0]

t = BPETokenizer(50000)
t.train_vocab(input_words)
t.vocab
```

```python
def gen_token_ids(bpe_encoded_tokens: dict[str,int]) -> dict[str,int]:
    res = [(k,bpe_encoded_tokens[k]) for k in bpe_encoded_tokens]
    res.sort(key = lambda pf : pf[1], reverse=True) # Decreasing sort of paris wrt frequence
    dout = dict()
    for i in range(len(res)):
        dout[res[i][0]] = i
    return dout

token_ids = gen_token_ids(t.vocab)
token_ids
# len(token_ids)
```

```python
def encode_words(words:list[str],bpe_token_ids: dict[str,int]) -> list[list[int]]:
    out = []
    for word in words:
        encoded_word = []
        for chari in range(len(word)):
            pair = word[chari] + EOS if chari == len(word)-1 else word[chari:chari+2]
            encoded_word.append( bpe_token_ids[pair])
        out.append(  encoded_word)
    return out
        
encoded_words_out = encode_words(input_words, token_ids)
encoded_words_out
```

```python
def max_input_length(encoded_words: list[list[int]]) -> int:
    lens = [len(ar) for ar in encoded_words]
    return max(lens) - 1 
    # -1 because, this is length of the input given to the model,
    # now model will generate a sequence of this length, starting from the second token of the sequence
max_input_length(encoded_words_out)    
```
